```python
age = int(input("Enter your age: "))

if age < 0:
    print("Invalid age.")
elif age <= 1:
    print("You are an Infant.")
elif age <= 3:
    print("You are a Toddler.")
elif age <= 12:
    print("You are a Child.")
elif age <= 19:
    print("You are a Teenager.")
elif age <= 39:
    print("You are a Young Adult.")
elif age <= 59:
    print("You are an Adult.")
elif age <= 79:
    print("You are a Senior Citizen.")
else:
    print("You are in the Elderly age group.")
```

```python
birth_year = int(input("Enter your birth year: "))

if 1981 <= birth_year <= 1996:
    print("You are a Millennial (Gen Y).")
elif 1997 <= birth_year <= 2012:
    print("You are Gen Z.")
elif 2013 <= birth_year:
    print("You are Gen Alpha.")
elif 1965 <= birth_year <= 1980:
    print("You are Gen X.")
elif birth_year < 1946:
    print("You belong to  generation.")
else:
    print("Invalid birth year.")
```

```python
year = int(input("Enter a year: "))

if (year % 400 == 0) or (year % 4 == 0 and year % 100 != 0):
    print(year, "is a Leap Year")
else:
    print(year, "is Not a Leap Year")
```

```python
num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))

op = input("Enter operation (+, -, *, /): ")

if op == "+":
    print("Result =", num1 + num2)

elif op == "-":
    print("Result =", num1 - num2)

elif op == "*":
    print("Result =", num1 * num2)

elif op == "/":
    if num2 != 0:
        print("Result =", num1 / num2)
    else:
        print("Division by zero is not allowed.")

else:
    print("Invalid operator.")
```
